## Uniform Cost Search (UCS)

```python
import heapq

def uniform_cost_search(graph, start, goal):
    """
    Implements the Uniform Cost Search algorithm.

    Args:
        graph (dict): A dictionary representing the graph where keys are nodes
                      and values are dictionaries of neighbors and their costs.
                      Example: {'A': {'B': 1, 'C': 10}, 'B': {'D': 2}, ...}
        start (str): The starting node.
        goal (str): The goal node.

    Returns:
        tuple: A tuple containing the path (list of nodes) and the total cost.
               Returns (None, None) if no path is found.
    """
    # A priority queue to store nodes to visit.
    # Each item is a tuple: (current_total_cost, current_node, path_to_current_node)
    # The priority queue automatically sorts by the first element (cost),
    # ensuring we always explore the path with the lowest cost first.
    priority_queue = [(0, start, [start])]

    # A dictionary to keep track of the lowest cost found so far to reach each node.
    # This helps in avoiding re-processing paths that are already known to be more expensive.
    # Initialize with the start node's cost as 0.
    visited_costs = {start: 0}

    # Continue as long as there are nodes to explore in the priority queue
    while priority_queue:
        # Get the node with the lowest cost from the priority queue
        # heapq.heappop removes and returns the smallest item.
        current_cost, current_node, current_path = heapq.heappop(priority_queue)

        # If we reached the goal, we've found the optimal path (due to how UCS works)
        if current_node == goal:
            return current_path, current_cost

        # Explore all neighbors of the current node
        # graph.get(current_node, {}) safely handles cases where a node might not have neighbors.
        for neighbor, edge_cost in graph.get(current_node, {}).items():
            # Calculate the total cost to reach this neighbor through the current path
            new_cost = current_cost + edge_cost

            # If this new path to the neighbor is cheaper than any previously found path,
            # or if this neighbor hasn't been visited yet:
            if neighbor not in visited_costs or new_cost < visited_costs[neighbor]:
                # Update the lowest cost to reach this neighbor
                visited_costs[neighbor] = new_cost
                # Create a new path by adding the neighbor to the current path
                new_path = current_path + [neighbor]
                # Add this neighbor to the priority queue to be explored later
                # The tuple is (total_cost_to_neighbor, neighbor_node, path_to_neighbor)
                heapq.heappush(priority_queue, (new_cost, neighbor, new_path))

    # If the loop finishes and the goal was never reached, no path exists.
    return None, None
```

### UCS Example Usage:

```python
# Define a simple sample graph for UCS
# Format: 'Node': {'Neighbor1': Cost1, 'Neighbor2': Cost2}
ucs_graph = {
    'A': {'B': 1, 'C': 3,'D':1},
    'B': {'E': 10},
    'C': {'F': 8},
    'D': {'G': 4},
    'G': {'H':1},
    'H':{},
    'E':{},
    'F':{}
}

start_node_ucs = 'A'
goal_node_ucs = 'H'

path_ucs, cost_ucs = uniform_cost_search(ucs_graph, start_node_ucs, goal_node_ucs)

if path_ucs:
    print(f"Uniform Cost Search Path from {start_node_ucs} to {goal_node_ucs}: {path_ucs}")
    print(f"Total Cost: {cost_ucs}")
else:
    print(f"No path found from {start_node_ucs} to {goal_node_ucs} using Uniform Cost Search.")
```

## A* Search Algorithm



```python
import heapq

def a_star_search(graph, start, goal, h):
    """
    Implements the A* Search algorithm.

    Args:
        graph (dict): A dictionary representing the graph where keys are nodes
                      and values are dictionaries of neighbors and their costs.
                      Example: {'A': {'B': 1, 'C': 10}, 'B': {'D': 2}, ...}
        start (str): The starting node.
        goal (str): The goal node.
        h (dict): A dictionary of heuristic values for each node, representing
                  the estimated cost from that node to the goal.
                  Example: {'A': 5, 'B': 3, ...}

    Returns:
        tuple: A tuple containing the path (list of nodes) and the total cost.
               Returns (None, None) if no path is found.
    """
    # The 'open_set' is a priority queue that stores nodes that are candidates for exploration.
    # Each item is a tuple: (f_score, node, g_score, path_to_node)
    # f_score = g_score + h(node). It's the estimated total cost from start to goal through this node.
    # g_score is the actual cost from the start node to the current node.
    # h(node) is the heuristic estimate from the current node to the goal.
    # The priority queue sorts by f_score.
    open_set = [(h[start], start, 0, [start])]

    # 'g_score' dictionary stores the lowest actual cost found so far to reach each node from the start.
    # Initialize all g_scores to infinity, except for the start node which is 0.
    g_score = {node: float('inf') for node in graph}
    g_score[start] = 0

    # Continue as long as there are nodes to explore
    while open_set:
        # Retrieve the node with the lowest f_score from the priority queue
        current_f, current_node, current_g, current_path = heapq.heappop(open_set)

        # If the current node is the goal, we have found the optimal path
        if current_node == goal:
            return current_path, current_g

        # Explore all neighbors of the current node
        for neighbor, edge_cost in graph.get(current_node, {}).items():
            # Calculate the cost to reach the neighbor through the current node
            tentative_g_score = current_g + edge_cost

            # If this new path to the neighbor is cheaper than any previously found path:
            # g_score.get(neighbor, float('inf')) safely retrieves the current g_score,
            # defaulting to infinity if the neighbor hasn't been reached yet.
            if tentative_g_score < g_score.get(neighbor, float('inf')):
                # Update the g_score for the neighbor with the new, lower cost
                g_score[neighbor] = tentative_g_score
                # Calculate the f_score for this new path to the neighbor
                # Use h.get(neighbor, float('inf')) for safety if a heuristic isn't defined for a node.
                new_f_score = tentative_g_score + h.get(neighbor, float('inf'))
                # Create a new path by extending the current path to include the neighbor
                new_path = current_path + [neighbor]
                # Add the neighbor to the open set for further exploration
                # (f_score, neighbor_node, g_score_to_neighbor, path_to_neighbor)
                heapq.heappush(open_set, (new_f_score, neighbor, tentative_g_score, new_path))

    # If the open set becomes empty and the goal was never reached, no path exists.
    return None, None
```

### A* Example Usage:

```python
# Define a simple sample graph for A*
# Format: 'Node': {'Neighbor1': Cost1, 'Neighbor2': Cost2}
a_star_graph = {
    'A': {'B': 1, 'C': 4},
    'B': {'C': 2, 'D': 3},
    'C': {'E': 3},
    'D': {'F':2,'G': 4},
    'E': {'G': 3},
    'F': {'G': 1},
    'G': {}
}

# Define heuristic values (estimated cost from each node to the goal 'G')
# These values should ideally never overestimate the actual cost.
a_star_heuristic = {
    'A': 5,
    'B': 6,
    'C': 4,
    'D': 3,
    'E': 3,
    'F': 1,
    'G': 0
}

start_node_astar = 'A'
goal_node_astar = 'G'

path_astar, cost_astar = a_star_search(a_star_graph, start_node_astar, goal_node_astar, a_star_heuristic)

if path_astar:
    print(f"A* Search Path from {start_node_astar} to {goal_node_astar}: {path_astar}")
    print(f"Total Cost: {cost_astar}")
else:
    print(f"No path found from {start_node_astar} to {goal_node_astar} using A* Search.")
```
