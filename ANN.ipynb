```python
import math

def sigmoid(x):
    return 1 / (1 + math.exp(-x))

def sigmoid_derivative(y):
    return y * (1 - y)

# Inputs and target output
x1 = 0.05
x2 = 0.10
target = 0.01

# Initial weights
w1 = 0.15  # x1 -> h1
w2 = 0.20  # x2 -> h1
w3 = 0.25  # x1 -> h2
w4 = 0.30  # x2 -> h2
w5 = 0.40  # h1 -> output
w6 = 0.45  # h2 -> output

learning_rate = 0.5

# -------- FORWARD PASS --------

net_h1 = x1*w1 + x2*w2
h1 = sigmoid(net_h1)

net_h2 = x1*w3 + x2*w4
h2 = sigmoid(net_h2)

net_o = h1*w5 + h2*w6
output = sigmoid(net_o)

print("----- FORWARD PASS -----")
print("Net input h1:", net_h1)
print("Output h1:", h1)
print("Net input h2:", net_h2)
print("Output h2:", h2)
print("Net input output neuron:", net_o)
print("Predicted output:", output)

# -------- ERROR --------

error = 0.5 * (target - output)**2
print("\n----- ERROR -----")
print("Target output:", target)
print("Error:", error)

# -------- BACKWARD PASS --------

delta_o = (target - output) * sigmoid_derivative(output)

delta_h1 = sigmoid_derivative(h1) * delta_o * w5
delta_h2 = sigmoid_derivative(h2) * delta_o * w6

print("\n----- BACKWARD PASS -----")
print("Output delta:", delta_o)
print("Hidden delta h1:", delta_h1)
print("Hidden delta h2:", delta_h2)

# -------- UPDATE WEIGHTS --------

w1 = w1 + learning_rate * delta_h1 * x1
w2 = w2 + learning_rate * delta_h1 * x2
w3 = w3 + learning_rate * delta_h2 * x1
w4 = w4 + learning_rate * delta_h2 * x2
w5 = w5 + learning_rate * delta_o * h1
w6 = w6 + learning_rate * delta_o * h2

print("\n----- UPDATED WEIGHTS -----")
print("w1 =", w1)
print("w2 =", w2)
print("w3 =", w3)
print("w4 =", w4)
print("w5 =", w5)
print("w6 =", w6)

# -------- FORWARD PASS AGAIN --------

net_h1_new = x1*w1 + x2*w2
h1_new = sigmoid(net_h1_new)

net_h2_new = x1*w3 + x2*w4
h2_new = sigmoid(net_h2_new)

net_o_new = h1_new*w5 + h2_new*w6
output_new = sigmoid(net_o_new)

# New error
new_error = 0.5 * (target - output_new)**2

print("\n----- NEW FORWARD PASS -----")
print("New net input h1:", net_h1_new)
print("New output h1:", h1_new)
print("New net input h2:", net_h2_new)
print("New output h2:", h2_new)
print("New net input output neuron:", net_o_new)
print("New predicted output:", output_new)
print("Target output:", target)
print("New error:", new_error)
```
